```python
from google.colab import userdata
from huggingface_hub import login
from transformers import AutoTokenizer
```

```python
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
```

```python
llama_model = "meta-llama/Llama-3.1-8B-Instruct"
ds_model = "deepseek-ai/DeepSeek-V3.1"
quen_mode = "Qwen/Qwen2.5-7B-Instruct"
phi_model = "microsoft/Phi-3-mini-4k-instruct"
```

```python
llama_tokenizer = AutoTokenizer.from_pretrained(llama_model, trust_remote_code=True)
ds_tokenizer = AutoTokenizer.from_pretrained(ds_model, trust_remote_code=True)
quen_tokenizer = AutoTokenizer.from_pretrained(quen_mode, trust_remote_code=True)
phi_tokenizer = AutoTokenizer.from_pretrained(phi_model, trust_remote_code=True)
```

```python
text = "I am curiously excited to see Hugging Face Tokenizers in action to my LLM engineers."
llama_tokens = llama_tokenizer.encode(text)
ds_tokens = ds_tokenizer.encode(text)
quen_tokens = quen_tokenizer.encode(text)
phi_tokens = phi_tokenizer.encode(text)

print(llama_tokens, ds_tokens, quen_tokens, phi_tokens, sep="\n====\n")
```

```python
llama_decoded = llama_tokenizer.batch_decode(llama_tokens)
ds_decoded = ds_tokenizer.batch_decode(ds_tokens)
quen_decoded = quen_tokenizer.batch_decode(quen_tokens)
phi_decoded = phi_tokenizer.batch_decode(phi_tokens)

print(llama_decoded, ds_decoded, quen_decoded, phi_decoded, sep="\n====\n")
```
